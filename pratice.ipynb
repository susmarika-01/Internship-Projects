```python
import spacy
```

```python
nlp=spacy.load("en_core_web_sm")
doc=nlp("My name is Susmarika. I'm 21 years old")

for sentence in doc.sents:
    print(sentence)
```

```python
for sentence in doc.sents:
    for word in sentence:
        print(word)
```
